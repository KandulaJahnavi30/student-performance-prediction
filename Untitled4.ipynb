```python
import pandas as pd

# Load the dataset
df = pd.read_csv("student_performance_5000.csv")

# Display the first 1000 rows
df.head(1000)
```

```python
# Check the number of rows and columns
print("Dataset Shape:")
print(df.shape)

# Check the column names
print("\nColumn Names:")
print(df.columns)

# Check the data types
print("\nData Types:")
print(df.dtypes)

# Check missing values
print("\nMissing Values:")
print(df.isnull().sum())
```

```python
print("Dataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns)

print("\nData Types:")
print(df.dtypes)

print("\nMissing Values:")
print(df.isnull().sum())
```

```python
# Check missing values
print("Missing values before cleaning:")
print(df.isnull().sum())

# Fill missing numerical values with the median
numeric_columns = df.select_dtypes(include=["int64", "float64"]).columns

for column in numeric_columns:
    df[column] = df[column].fillna(df[column].median())

# Fill missing text/categorical values with the most common value
categorical_columns = df.select_dtypes(include=["object"]).columns

for column in categorical_columns:
    df[column] = df[column].fillna(df[column].mode()[0])

# Check again
print("\nMissing values after cleaning:")
print(df.isnull().sum())
```

```python
# Convert categorical input columns into numbers

df = pd.get_dummies(
    df,
    columns=[
        "Extracurricular_Activities",
        "Internet_Access",
        "Parent_Education"
    ],
    drop_first=True
)

# Display the updated dataset
print("New dataset shape:")
print(df.shape)

print("\nUpdated columns:")
print(df.columns.tolist())
```

```python
# Remove Student_ID because it is only an identifier
df = df.drop("Student_ID", axis=1)

# Features (input columns)
X = df.drop("Performance", axis=1)

# Target (output column)
y = df["Performance"]

print("Features (X):")
print(X.columns.tolist())

print("\nTarget (y):")
print(y.name)

print("\nShape of X:", X.shape)
print("Shape of y:", y.shape)
```

```python
from sklearn.model_selection import train_test_split

# Split the dataset
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

# Check the sizes
print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
print("Training target:", y_train.shape)
print("Testing target:", y_test.shape)
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, classification_report

# Create the Decision Tree model
decision_tree = DecisionTreeClassifier(
    random_state=42
)

# Train the model
decision_tree.fit(X_train, y_train)

# Make predictions
y_pred_dt = decision_tree.predict(X_test)

# Calculate accuracy
accuracy_dt = accuracy_score(y_test, y_pred_dt)

print("Decision Tree Accuracy:", accuracy_dt)

# Detailed performance report
print("\nClassification Report:")
print(classification_report(y_test, y_pred_dt))
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report

# Create the Random Forest model
random_forest = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

# Train the model
random_forest.fit(X_train, y_train)

# Make predictions
y_pred_rf = random_forest.predict(X_test)

# Calculate accuracy
accuracy_rf = accuracy_score(y_test, y_pred_rf)

print("Random Forest Accuracy:", accuracy_rf)

# Detailed performance report
print("\nClassification Report:")
print(classification_report(y_test, y_pred_rf))
```

```python
# Features for regression
X_reg = df.drop(["Final_Score", "Performance"], axis=1)

# Target for regression
y_reg = df["Final_Score"]

print("Features shape:", X_reg.shape)
print("Target shape:", y_reg.shape)
```

```python
from sklearn.model_selection import train_test_split

X_train_reg, X_test_reg, y_train_reg, y_test_reg = train_test_split(
    X_reg,
    y_reg,
    test_size=0.20,
    random_state=42
)

print("Training data:", X_train_reg.shape)
print("Testing data:", X_test_reg.shape)
```

```python
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

# Create the Regression Tree
regression_tree = DecisionTreeRegressor(
    random_state=42
)

# Train the model
regression_tree.fit(X_train_reg, y_train_reg)

# Make predictions
y_pred_reg = regression_tree.predict(X_test_reg)

# Calculate evaluation metrics
mae = mean_absolute_error(y_test_reg, y_pred_reg)
mse = mean_squared_error(y_test_reg, y_pred_reg)
rmse = np.sqrt(mse)
r2 = r2_score(y_test_reg, y_pred_reg)

print("Regression Tree Results")
print("-----------------------")
print("MAE :", mae)
print("MSE :", mse)
print("RMSE:", rmse)
print("R²  :", r2)
```

```python
# Create a comparison table

model_comparison = pd.DataFrame({
    "Model": [
        "Decision Tree Classifier",
        "Random Forest Classifier",
        "Decision Tree Regressor"
    ],
    "Metric": [
        "Accuracy",
        "Accuracy",
        "R² Score"
    ],
    "Score": [
        accuracy_dt,
        accuracy_rf,
        r2
    ]
})

print("MODEL COMPARISON")
print("================")
display(model_comparison)
```

```python
# Final results of the three models

print("===== FINAL MODEL RESULTS =====")

print("\n1. Decision Tree Classifier")
print("Accuracy:", accuracy_dt)

print("\n2. Random Forest Classifier")
print("Accuracy:", accuracy_rf)

print("\n3. Regression Tree")
print("MAE:", mae)
print("RMSE:", rmse)
print("R² Score:", r2)
```
